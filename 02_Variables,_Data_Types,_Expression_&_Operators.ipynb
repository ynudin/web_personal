## Casting Tipe Variabel

```python
# INTEGER
data = -10
print("value:",data,"type:",type(data))

data_float = float(data)
print("nilai: ", data_float, ", tipe: ", type(data_float)) #akan ada angka .0(desimal)

data_str = str(data)
print("nilai: ", data_str, ", tipe: ", type(data_str))

data_bool = bool(data)
print("nilai: ", data_bool, ", tipe: ", type(data_bool)) # semua nilai integer selain 0 bernilai False
```

```python
# FLOAT
data = -0.0
print("value:",data,"type:",type(data))

data_int = int(data)
print("nilai: ", data_int, ", tipe: ", type(data_int))

data_str = str(data)
print("nilai: ", data_str, ", tipe: ", type(data_str))

data_bool = bool(data)
print("nilai: ", data_bool, ", tipe: ", type(data_bool)) # semua nilai integer selain 0 bernilai False
```

```python
# String
data = "10"
print("value:",data,"type:",type(data))

data_int = int(data)
print("nilai: ", data_int, ", tipe: ", type(data_int)) # Untuk string angka nilai desimal seperti 10.0 tidak bisa diubah ke tipe data integer(terdapat ".")

data_float = float(data) #string selain angka dan koma(.) tidak dapat diubah ke tipe data float
print("nilai: ", data_float, ", tipe: ", type(data_float))

data_bool = bool(data)
print("nilai: ", data_bool, ", tipe: ", type(data_bool)) # selain string kosong "" maka bernilai True
```

```python
# BOOLEAN
data = False
print("value:",data,"type:",type(data))

data_int = int(data)
print("nilai: ", data_int, ", tipe: ", type(data_int)) # Jika True maka nilai 1, False maka 0
data_str = str(data)
print("nilai: ", data_str, ", tipe: ", type(data_str))
data_float = float(data)
print("nilai: ", data_float, ", tipe: ", type(data_float)) # Jika True maka nilai 1.0, False maka 0.0
```
